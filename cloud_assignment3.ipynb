```python
# Import Libraries & Set Seed
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(5)
```

```python
# Generate random sales data (12 months × 4 categories)
sales_data = np.random.randint(1000, 5000, size=(12, 4))
print("Sales Data:\n", sales_data)
```

```python
#Convert to DataFrame
columns = ["Electronics", "Clothing", "Home & Kitchen", "Sports"]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
df = pd.DataFrame(sales_data, columns=columns, index=months)

print(df)
```

```python
print(df.head())
print(df.describe())
```

```python
# Total sales per category
category_total = df.sum()
print("Total per category:\n", category_total)

# Total sales per month
monthly_total = df.sum(axis=1)
print("Total per month:\n", monthly_total)
```

```python
# Calculate growth between months
growth = df.pct_change()

# Average growth per category
avg_growth = growth.mean()

print("Average Growth:\n", avg_growth)
```

```python
# Add Total Sales column
df["Total Sales"] = df.sum(axis=1)

# Add Growth Rate column
df["Growth Rate"] = df["Total Sales"].pct_change()

print(df)
```

```python
# Apply 15% discount on Clothing
df["Clothing"] = df["Clothing"] * 0.85

print("After Discount:\n", df)
```

```python
# Line plot for monthly sales
df[["Electronics", "Clothing", "Home & Kitchen", "Sports"]].plot(marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
# Boxplot
sns.boxplot(data=df[["Electronics", "Clothing", "Home & Kitchen", "Sports"]])

plt.title("Sales Distribution")
plt.show()
```

```python
# Q2: NumPy Operations
array = np.array([[1, -2, 3], [-4, 5, -6]])

# i. Absolute value
print("Absolute:\n", np.abs(array))

# ii. Percentiles
flat = array.flatten()
print("Percentiles (flattened):", np.percentile(flat, [25, 50, 75]))

print("Column-wise percentiles:\n", np.percentile(array, [25,50,75], axis=0))
print("Row-wise percentiles:\n", np.percentile(array, [25,50,75], axis=1))

# iii. Mean, Median, Std
print("Mean:", np.mean(flat))
print("Median:", np.median(flat))
print("Std Dev:", np.std(flat))

print("Column-wise Mean:", np.mean(array, axis=0))
print("Row-wise Mean:", np.mean(array, axis=1))
```

```python
# Q3: Floor, Ceil, Truncate
a = np.array([-1.8, -1.6, -0.5, 0.5, 1.6, 1.8, 3.0])

print("Floor:", np.floor(a))
print("Ceil:", np.ceil(a))
print("Truncate:", np.trunc(a))
print("Rounded:", np.round(a))
```

```python
# Q4: Swap Using Temporary Variable

lst = [10, 20, 30, 40]
temp = lst[0]
lst[0] = lst[2]
lst[2] = temp

print("Swapped List:", lst)
```

```python
#Q5: Swap in Set via List

s = {1, 2, 3, 4}
lst = list(s)
lst[0], lst[1] = lst[1], lst[0]
s = set(lst)
print("Updated Set:", s)
```
